```python
# OpenAI style API call using OpenAI SDK
# !pip install openai
from openai import OpenAI

API_KEY = "dummy_api_key"

client = OpenAI(
    # lmstudio API
    base_url="http://localhost:1234/v1",
    api_key=API_KEY,
)
```

```python
# list available models
models = client.models.list()
for m in models.data: print(m.id)
```

```python
# get a response from a model
response = client.responses.create(
    model="google/gemma-4-e4b",
    input="What is the capital of India?"
)
print(response.output_text)
```

```python
print(response)
```

```python
# temperature, max_output_tokens parameters
response = client.responses.create(
    model="google/gemma-4-e4b",
    input="Write a short poem about the sun.",
    temperature=2.0,
    max_output_tokens=1000
)
print(response.output_text)
```

```python
# create embeddings
response = client.embeddings.create(
    model="text-embedding-nomic-embed-text-v1.5",
    input="Hello world"
)
print("embedding:", response.data[0].embedding)
print("length:", len(response.data[0].embedding))
```

```python
# Without the SDK
import requests

# OpenAI style API call to lmstudio API
response = requests.post(
    url="http://localhost:1234/v1/chat/completions",
    headers={
        "Content-Type": "application/json",
        "Authorization": f"Bearer {API_KEY}"
    },
    json={
        "model": "google/gemma-4-e4b",
        "messages": [
            {"role": "user", "content": "What is the capital of India?"}
        ],
        "temperature": 0.5,
        "max_tokens": 1024
    },
    timeout=60
)

response.raise_for_status()
print(response.json()["choices"][0]["message"]["content"])
```
